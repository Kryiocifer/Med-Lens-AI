```python
import os
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

# Optional: for future use
import torch
from torchvision import transforms
```

```python
# Path to your brain tumor dataset
data_path = r"Z:\Misc Data\College\Med Lens AI\data\raw\brain_tumor_mri"

# Check if path exists
print("Path exists:", os.path.exists(data_path))
print("Contents:", os.listdir(data_path))
```

```python
# Define classes
classes = ['glioma', 'meningioma', 'notumor', 'pituitary']

# Count images in Training and Testing
for split in ['Training', 'Testing']:
    print(f"\n=== {split} ===")
    split_path = os.path.join(data_path, split)
    
    for cls in classes:
        class_path = os.path.join(split_path, cls)
        if os.path.exists(class_path):
            num_images = len(os.listdir(class_path))
            print(f"{cls}: {num_images} images")
        else:
            print(f"{cls}: Folder not found")
```

```python
def show_sample_images(data_path, split='Training', num_samples=3):
    classes = ['glioma', 'meningioma', 'notumor', 'pituitary']
    
    fig, axes = plt.subplots(len(classes), num_samples, figsize=(12, 10))
    fig.suptitle(f'Sample Images from {split} Set', fontsize=16)
    
    for i, cls in enumerate(classes):
        class_path = os.path.join(data_path, split, cls)
        images = os.listdir(class_path)[:num_samples]
        
        for j, img_name in enumerate(images):
            img_path = os.path.join(class_path, img_name)
            img = Image.open(img_path).convert('RGB')
            
            axes[i, j].imshow(img)
            axes[i, j].set_title(f"{cls}")
            axes[i, j].axis('off')
    
    plt.tight_layout()
    plt.show()

# Show samples from Training set
show_sample_images(data_path, split='Training', num_samples=3)
```
